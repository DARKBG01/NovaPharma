```python
#!/usr/bin/env python3
# -*- coding: utf-8 -*-
"""
migrate_ui.py — Migre les icônes de MainWindow.ui
vers la nouvelle bibliothèque svg_gen.py
"""
from pathlib import Path
import re

# ─── Mapping ancien → nouveau ────────────────────────────────────
ICON_MAP = {
    # Logo / sidebar
    "pharmacy-icon-svgrepo-com2.svg":  "pharmacy.svg",

    # Navigation
    "home-svgrepo-com.svg":            "home.svg",
    "home-svgrepo-com2.svg":           "home.svg",
    "home-svgrepo-com3.svg":           "home.svg",
    "pie-chart-1-svgrepo-com.svg":     "chart-pie.svg",
    "pie-chart-1-svgrepo-com2.svg":    "chart-pie.svg",
    "pie-chart-1-svgrepo-com2 (Copie).svg": "chart-pie.svg",
    "medical-medicine-health-11-svgrepo-com.svg":       "pill.svg",
    "medical-medicine-health-11-svgrepo-com2.svg":      "pill.svg",
    "medical-medicine-health-11-svgrepo-com2 (Copie).svg": "pill.svg",
    "users-group-svgrepo-com3.svg":    "users.svg",
    "users-group-svgrepo-com4.svg":    "users.svg",
    "users-group-svgrepo-com4 (Copie).svg": "users.svg",
    "invoice-dollar-svgrepo-com.svg":  "invoice.svg",
    "invoice-dollar-svgrepo-com2.svg": "invoice.svg",
    "invoice-dollar-svgrepo-com (Copie).svg": "invoice.svg",
    "history-svgrepo-com.svg":         "history.svg",
    "history-svgrepo-com2.svg":        "history.svg",
    "history-svgrepo-com3.svg":        "history.svg",
    "report-svgrepo-com.svg":          "chart-line.svg",
    "report-svgrepo-com2.svg":         "chart-line.svg",
    "report-svgrepo-com (Copie).svg":  "chart-line.svg",
    "user-setting-svgrepo-com.svg":    "settings.svg",
    "user-setting-svgrepo-com2.svg":   "settings.svg",
    "user-setting-svgrepo-com (Copie).svg": "settings.svg",

    # Header / stats
    "currency-dollar-circle-svgrepo-com.svg":  "dollar.svg",
    "shopping-bag-svgrepo-com.svg":            "shopping-bag.svg",
    "user-plus-alt-1-svgrepo-com.svg":         "user-plus.svg",
    "biohazard-svgrepo-com.svg":               "alert-triangle.svg",

    # Boutons
    "exclamation-mark-triangle-f-svgrepo-com.svg": "alert-triangle.svg",
    "add-svgrepo-com.svg":                     "plus.svg",
    "edit-svgrepo-com (2).svg":                "edit.svg",
    "edit-svgrepo-com.svg":                    "edit.svg",
    "search-alt-2-svgrepo-com.svg":            "search.svg",
    "shopping-cart-free-5-svgrepo-com.svg":    "shopping-cart.svg",
}


def migrate(input_path: str, output_path: str) -> None:
    text = Path(input_path).read_text(encoding="utf-8")

    # Compter les remplacements
    stats = {}
    for old, new in ICON_MAP.items():
        # Échappe les caractères spéciaux (parenthèses, espaces, etc.)
        pattern = re.escape(old)
        occurrences = len(re.findall(pattern, text))
        if occurrences:
            text = re.sub(pattern, new, text)
            stats[old] = (new, occurrences)

    Path(output_path).write_text(text, encoding="utf-8")

    # Rapport
    print(f"✅ Fichier migré : {output_path}\n")
    if stats:
        print(f"{'ANCIEN':<55} → {'NOUVEAU':<25} {'#':>4}")
        print("-" * 90)
        for old, (new, n) in sorted(stats.items()):
            print(f"{old:<55} → {new:<25} {n:>4}")
    else:
        print("⚠️  Aucun remplacement effectué. Vérifie le chemin.")


if __name__ == "__main__":
    migrate("Assets/Ui/Interface_main.ui", "Assets/Ui/MainWindow_new.ui")
    print("\n→ Renomme MainWindow_new.ui en MainWindow.ui après vérification.")
```
